# USING YOLO

Yolo is a Computer Vision library that identifies and localizates objects in an image, fortunately the hardwork was done before in the boxing task at [roboflow](roboflow.com), now all we have to do it's to train our model by ourselves.

![Image](../LOGSBIN/assets/doom_attack_57_jpg.rf.aface71250b423830e724ec10f18e0ef.jpg)
>[!NOTE]
>Notice that YOLO uses an image detection resolution of 448x448, that's why the image looks like that, the boxing is going to be found at labels file

```python
import os
from dotenv import load_dotenv

load_dotenv()

api_key = os.getenv("ROBOFLOW_API_KEY")
workspace = os.getenv("WORKSPACE")
```

```python
from roboflow import Roboflow
rf = Roboflow(api_key=api_key)
project = rf.workspace(workspace).project("doom-enemy-classifier")
dataset = project.version(1).download("yolov11")
```

### Here comes the real training 
A YOLO model only need a few lines and a little time to get the model trained and we are going to use a public chekpoint to make the training faster.

```python
from ultralytics import YOLO

model = YOLO("yolo11n.pt")

results = model.train(
    data = f"{dataset.location}/data.yaml",
    epochs = 30,
    imgsz = 640,
    cache = True,
    device = "mps" #IMPORTANT: If you are using a Nvidia GPU just comment this kwarg
)
```

## Results:

The model over all didn't reached a crazy 10/10 classification but reached a 0.864 mAP50 which means that the prediction box over the enemy had at least average precision of .50.


### It's going to be always precise?
No, definitely no, but it's a great aproach for only 290 labeled images, wehere some classes were underrepresented over others.
